Conexão do banco de dados com o jupyter

```python
from config import DB_CONFIG
import mysql.connector

conn = mysql.connector.connect(**DB_CONFIG)
```

```python
import pandas as pd
```

```python
df_exp = pd.read_sql("""
SELECT 
    experience_years,
    AVG(salary) as avg_salary
FROM salary_data
GROUP BY experience_years
ORDER BY experience_years
""", conn)
```

```python
import matplotlib.pyplot as plt

plt.figure()

plt.plot(df_exp["experience_years"], df_exp["avg_salary"])

plt.title("Salário médio por anos de experiência")
plt.xlabel("Anos de experiência")
plt.ylabel("Salário médio anual em dólar")

plt.grid()
plt.show()
```

```python
df_exp1 = pd.read_sql("""
SELECT 
    education_level,
    AVG(salary) as avg_salary
FROM salary_data
GROUP BY education_level
ORDER BY avg_salary DESC
""", conn)
```

```python
plt.figure()

plt.plot(df_exp1["education_level"], df_exp1["avg_salary"])

plt.title("Salário médio por nível de escolaridade")
plt.xlabel("Nível de escolaridade")
plt.ylabel("Salário médio anual em dólar")

plt.grid()
plt.show()
```

```python
df_exp2 = pd.read_sql("""
SELECT 
    remote_work,
    AVG(salary)
FROM salary_data
GROUP BY remote_work;
""", conn)
```

```python
plt.figure()

plt.bar(df_exp2["remote_work"], df_exp2["AVG(salary)"], color="darkgreen")

plt.title("Salário médio por tipo de trabalho")
plt.xlabel("Trabalho remoto")
plt.ylabel("Salário médio")

plt.tight_layout()
plt.show()
```
